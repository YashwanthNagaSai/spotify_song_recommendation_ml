```python
X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=features,
    index=df.index
)
print(X_scaled_df)
```

```python
from sklearn.cluster import KMeans

inertia = []

for k in range(2,11):
    kmeans = KMeans(n_clusters = k, random_state = 42, n_init = 10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize = (8, 5))
plt.plot(range(2, 11), inertia, marker = 'o')
plt.xlabel("K Values")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters = k, random_state = 42, n_init = 10)
    Labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, Labels)
    silhouette_scores.append(score)

plt.figure(figsize = (8, 5))
plt.plot(range(2, 11), silhouette_scores, marker = 'o')
plt.xlabel("K_Values")
plt.ylabel("Silhouette_Score")
plt.title("Silhouette _Method")
plt.show()
```

```python
from sklearn.metrics import calinski_harabasz_score, davies_bouldin_score

calinski_scores = []
davies_bouldin_scores = []
results = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters = k, random_state = 42, n_init = 10)
    Labels = kmeans.fit_predict(X_scaled)
    silhouette = silhouette_score(X_scaled, Labels)
    calinski_score = calinski_harabasz_score(X_scaled, Labels)
    Score = davies_bouldin_score(X_scaled, Labels)
    calinski_scores.append(calinski_score)
    davies_bouldin_scores.append(Score)
    results.append({
        "k": k, 
        "Silhouette_Score" : silhouette, 
        "Calinski_Harabasz_Score" : calinski_score, 
        "Davies_Bouldin_Score" : Score
    })

results_pf = pd.DataFrame(results)
print(results_pf)

plt.figure(figsize = (8, 5))
plt.plot(range(2, 11), davies_bouldin_scores, marker = 'o')
plt.xlabel("K Values")
plt.ylabel("Davies-Bouldin Score")
plt.title("Davies-Bouldin Method")
plt.show()

plt.figure(figsize = (8, 5))
plt.plot(range(2, 11), calinski_scores, marker = 'o')
plt.xlabel("K Values")
plt.ylabel("Calinski-Harabasz Score")
plt.title("Calinski-Harabasz Method")
plt.show()
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

df["cluster"] = kmeans.fit_predict(X_scaled)

print(df["cluster"].value_counts().sort_index())

cluster_profile = df.groupby("cluster")[features].mean()

print(cluster_profile.round(3))
```

```python
from sklearn.decomposition import PCA

# PCA for visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

cluster_names = {
    0: "Acoustic / Low-Energy",
    1: "High-Energy / Fast-Paced",
    2: "Instrumental / Energetic",
    3: "Danceable / Speech-Rap",
    4: "Danceable / Upbeat-Happy"
}

plt.figure(figsize=(20, 10))

# Plot each cluster separately
for cluster in range(5):
    cluster_points = X_pca[df["cluster"] == cluster]

    plt.scatter(
        cluster_points[:, 0],
        cluster_points[:, 1],
        label=cluster_names.get(cluster, f"Cluster {cluster}"),
        alpha=0.6
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering with PCA (K=5)")

# Show which color represents which cluster
plt.legend(title="Clusters")

plt.show()
```

```python
from sklearn.metrics.pairwise import euclidean_distances

def recommend_similar_songs(song_name, n=5):

    # Find matching songs
    selected_song_df = df[
        df['track_name'].str.contains(
            song_name,
            case=False,
            na=False
        )
    ]

    # Check if song exists
    if selected_song_df.empty:
        return "Song not found in the dataset!"

    # Select the first matching song
    selected_index = selected_song_df.index[0]

    # Selected song details
    selected_cluster = df.loc[selected_index, 'cluster']
    selected_track_name = df.loc[selected_index, 'track_name']
    selected_artist = df.loc[selected_index, 'track_artist']

    # Get other songs from the same cluster
    # Exclude the same song + artist
    cluster_indices = df[
        (df['cluster'] == selected_cluster) &
        ~(
            (df['track_name'] == selected_track_name) &
            (df['track_artist'] == selected_artist)
        )
    ].index

    # Selected song features
    selected_features = X_scaled_df.loc[[selected_index]]

    # Other songs' features
    cluster_features = X_scaled_df.loc[cluster_indices]

    # Calculate Euclidean distances
    distances = euclidean_distances(
        selected_features,
        cluster_features
    )[0]

    # Create recommendations
    recommendations = df.loc[cluster_indices].copy()

    recommendations['distance'] = distances

    # Sort by similarity
    recommendations = recommendations.sort_values(
        by='distance'
    )

    # Remove duplicate song + artist combinations
    recommendations = recommendations.drop_duplicates(
        subset=['track_name', 'track_artist']
    )

    # Return Top N
    return recommendations[
        [
            'track_name',
            'track_artist',
            'playlist_genre',
            'distance'
        ]
    ].head(n)
```
